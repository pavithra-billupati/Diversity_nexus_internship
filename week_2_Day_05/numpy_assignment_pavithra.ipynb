Part G – Main Hands-on Activity: Student Marks Analyzer
Create a NumPy-based program to analyze student marks. Use this data:
marks = np.array([[78, 85, 92], [66, 74, 81], [90, 88, 95], [55, 69, 72]])
Each row represents one student; each column represents a subject.
Your program should:
Print the full marks array, its dimensions, and its shape.
Print the marks of the first student.
Print the marks for the second subject for all students.
Add 5 bonus marks to every score using a vectorized operation.
Cap adjusted marks at 100 using np.minimum or an equivalent NumPy operation.
Calculate each student's average marks using an axis argument.
Find the highest mark in the full array.
Select all adjusted scores that are 80 or above.
Print results with meaningful labels.

```python
import numpy as np

marks = np.array([
    [78, 85, 92],
    [66, 74, 81],
    [90, 88, 95],
    [55, 69, 72]
])

print("Marks Array:\n", marks)
print("Dimensions:", marks.ndim)
print("Shape:", marks.shape)

print("First Student Marks:", marks[0])
print("Second Subject Marks:", marks[:, 1])

adjusted_marks = np.minimum(marks + 5, 100)
print("Adjusted Marks:\n", adjusted_marks)

averages = np.mean(adjusted_marks, axis=1)
print("Student Averages:", averages)

print("Highest Mark:", np.max(marks))

print("Adjusted Scores >= 80:", adjusted_marks[adjusted_marks >= 80])
```

1. What is NumPy, and why is it useful?
Answer:NumPy (Numerical Python) is a Python library used to work with arrays and perform mathematical calculations efficiently. 
2. What is a NumPy array?
Answer: A NumPy array is a data structure used to store multiple elements in one or more dimensions.
3. What is the difference between a Python list and a NumPy array?
Answer: A Python list stores multiple items and can contain different data types, while a NumPy array stores elements efficiently and supports fast mathematical operations.
4. What do ndim, shape, size, and dtype represent?
Answer:- ndim: Number of dimensions.
 shape: Size along each dimension.
 size: Total number of elements.
 dtype: Data type of the elements. 
5. What is indexing? What is negative indexing?
Answer:Indexing: Indexing is used to access an element in an array using its position, starting from index 0.
Negative Indexing: Negative indexing is used to access elements in an array from the end, starting with index -1.
6. What is slicing?
Answer: Slicing selects a range of elements from an array using start, stop, and step values.
7. What is a vectorized operation?
Answer: A vectorized operation performs an operation on all elements of a NumPy array at once without using an explicit loop.
8. Why must arrays generally have compatible shapes for element-wise arithmetic?
Answer: Arrays must have compatible shapes for element-wise arithmetic so that their elements can be matched correctly for calculations.
