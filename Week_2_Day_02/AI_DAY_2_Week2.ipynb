```python
#PART-A
student = {"name": "Pavithra", "age": 21, "course": "Python", "marks": 85}

#Print the complete dictionary.
print(student)
#Print the student's name
print(student["name"])
#Print the student's marks
print(student["marks"])
#Change the marks to 90
student["marks"]=90
#add a city key
student["city"]="Nellore"
print(student)
#remove the age key
del student["age"]
key=student.pop("city")
print(key)
print(student)
```

Why is a dictionary useful for storing information about a student?

A dictionary is useful for storing student information because it stores data using keys. Each key helps us identify what the corresponding value represents.


```python
#PART B
# Practice: keys(), values(), items(), get(), update(), pop()
# Print all keys.
print(student.keys())
# Print all values.
print(student.values())
# Print all key-value pairs using items().
for key,value in student.items():
    print(f"{key} : {value}")
# Use get() to safely access a key.
print(student.get("city"))
print(student.get("name"))
print(student.get("city","not availble"))
# Use update() to add or change information.
student.update(city="Nellore")
student.update(age=20)
print(student)
student.update(age=21)
print(student)
# Use pop() to remove one item.
student.pop("marks")
print(student)
```

```python
#PART C
students = {
    "student1": {"name": "Rahul", "marks": 85},
    "student2": {"name": "Priya", "marks": 92},
    "student3": {"name": "Arjun", "marks": 78}
    }
# Print the complete dictionary.
print(students)
# Print Priya's name.
print(students["student2"]["name"])
# Print Priya's marks.
print(students["student2"]["marks"])
# Change Arjun's marks to 82.
students["student3"] ["marks"]=80
print(students)
# Add a city for each student.
students["student1"] ["city"]="Nellore"
students["student2"] ["city"]="giddalur"
students["student3"] ["city"]="tirupati"
print(students)
# Print each student's name and marks using a loop.
for student in students.values():
    for key,value in student.items():
        if key=="name" or key=="marks":
            print(key, " : ", value)    
#Find the student with the highest marks.
mark=0
name=""
for student in students.values():
    for key,value in student.items():
        if key=="marks":
            if mark<=value:
                mark=value
                name=student["name"]
print(name,"got highest marks",mark)
    
```

```python
#PART D
from collections import Counter
fruits = ["apple", "banana", "apple", "orange", "banana", "apple"]
result=Counter(fruits)
print(result)
# Create a Counter for fruits.
count=0
for i in fruits:
    count +=1
print(f"{count} fruits ")
# Find the count of apple.
count_apple=0
for i in fruits:
    if i=="apple":
        count_apple +=1
print(f"{count_apple} apple in fruits ")
# Find the count of banana.
count_banana=0
for i in fruits:
    if i=="banana":
        count_banana+=1
print(f"{count_banana} apple in fruits ")
# Find the most common fruit.
countf={}
most=""
cou=0
for i in fruits:
    if i in countf:
        countf[i] +=1
    else:
        countf[i] = 1
print(countf)
for key,value in countf.items():
    if cou<value:
        cou=value
        most=key
print(f"the most common fruit is {most} ")

# Try Counter with the string "programming".
count_s=0
for i in "programming":
    count_s +=1
print(f"{count_s} in programming")
```

```python
#PART H
student = {"name": "Rahul", "age": 21, "course": "Python", "marks": 85}
# Your program should:
# Display student details.
print(student)
# Check whether the student passed or failed.
if student["marks"]>45:
    print("pass")
else:
    print("fail")
# Assign a grade based on marks.
if student["marks"]>=45 and student["marks"]<=59:
    print("C")
elif student["marks"]>=60 and student["marks"]<=89:
    print("B")
elif student["marks"]>=90 and student["marks"]<=100:
    print("A")
# Allow marks to be updated.
student["marks"]=int(input("Update the marks:"))

# Display the final student information.
print(student)

# Bonus: Store 3–5 students in a nested dictionary and find the highest scorer.
students = {
    "student1": {"name": "Rahul", "marks": 85},
    "student2": {"name": "Priya", "marks": 92},
    "student3": {"name": "Arjun", "marks": 78}
    }
highest=0
topper_name=""
for student in students.values():
    if student["marks"]>highest:
        highest=student["marks"]
        topper_name=student["name"]
print(f" the highest score {highest} by {topper_name}")
```

```python
#PART H
contacts = {"Rahul": "9876543210", "Priya": "9123456780", "Arjun": "9988776655"}
# Add a new contact.
contacts.update({"Pavithra":"2345678910"})
# Search for a contact by name.
# Update a phone number.
name=input("Enter name :")
if name in contacts:
    print("phone number :",contacts[name])
else:
    number=int(input("enter number :"))
    contacts.update({name : number})
print(contacts)
# Delete a contact.
del_con=input("Enter name:")
if del_con in contacts:
    del contacts[del_con]
print(contacts)
# Display all contacts.
print(contacts)
# Bonus: Store phone, email, and city for each contact using a nested dictionar

contacts_stu = {}
for i in range(2):
    name=input("Enter the name:")
    phone=input("Enter the phone :")
    email=input("enter the email :")
    city=input("enter city:")
    contacts_stu[name]={
        "phone":phone,
        "email":email,
        "city":city
    }
print(contacts_stu)

```

```python
from collections import Counter
sentence="Python is easy and Python is powerful"
# Bonus: convert all words to lowercase before counting.
senten=sentence.lower()
print(senten)
# Convert the sentence into words.
sen=senten.split()
print(sen)
# Use Counter to count each word.
count=Counter(sen)
# Display the word counts.
print(count)
# Find the most frequently used word.
most=0
word=""
for key,value in count.items():
    if most<value:
        most=value
        word=key
print(f"most commonly used {word} in {most}")
print(count.most_common(1))
```
