```python
import pandas as pd

df = pd.read_csv("student.csv")
high_marks = df[df["Marks"] > 80]
print(high_marks)
combined = df[(df["Marks"] > 70) & (df["Age"] >= 21)]
print(combined)
result = df[(df["Marks"] > 90) | (df["Age"] < 21)]
print(result)
selected=df[df["Branch"]=="CSE"]
print(selected)
result=df[df["Marks"]>75][["Name","Marks"]]
print(result)
print("Marks above 80:", df[df["Marks"] > 80].shape[0])
print("CSE students:", df[df["Branch"] == "CSE"].shape[0])
print("Marks above 70 and age >= 21:",df[(df["Marks"] > 70) & (df["Age"] >= 21)].shape[0])

df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])

date_result = df[
    (df["Joining_Date"] >= "2025-06-10") &
    (df["Joining_Date"] <= "2025-06-20")
]

print(date_result)
```

 Part E – Data Quality Check
Check missing values using isna().sum().
Check duplicate rows using duplicated().sum().
Inspect data types and identify columns that should be numeric or dates.
If dates exist, convert them using pd.to_datetime() where appropriate.
Do not remove missing values or duplicates automatically; first explain what you found.

```python

import pandas as pd

df = pd.read_csv("student.csv")

# 1. Check missing values
print("Missing values:")
print(df.isna().sum())

# 2. Check duplicate rows
print("\nDuplicate rows:")
print(df.duplicated().sum())

# 3. Inspect data types
print("\nData types:")
print(df.dtypes)

# 4. Convert date column if it exists
if "Joining_Date" in df.columns:
    df["Joining_Date"] = pd.to_datetime(
        df["Joining_Date"], errors="coerce"
    )
    print("\nUpdated data types:")
    print(df.dtypes)

# 5. Report findings without deleting data
print("\nData quality summary:")
print("Total rows:", len(df))
print("Total columns:", len(df.columns))
print("Total missing values:", df.isna().sum().sum())
print("Total duplicate rows:", df.duplicated().sum())
```
