```python
# line plot - shows changes over time
import matplotlib.pyplot as plt
months = ['Jan', 'Feb', 'Mar', 'Apr']
sales = [100, 150, 130, 180]
plt.plot(months, sales, marker='o')
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.show()
```

```python
# bar chart - compares different categories
import matplotlib.pyplot as plt
products = ['Books', 'Pens', 'Bags']
sales = [50, 80, 40]
plt.bar(products, sales)
plt.title('Product Sales')
plt.xlabel('Product')
plt.ylabel('Sales')
plt.show()
```

```python
# histogram - shows how often values occur
import matplotlib.pyplot as plt
marks = [40, 45, 50, 50, 55, 60, 60, 65, 70, 80]
plt.hist(marks, bins=5, edgecolor='black')
plt.title('Distribution of Marks')
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.show()
```

```python
# scatter plot - shows the relationship between two values
import matplotlib.pyplot as plt
hours = [1, 2, 3, 4, 5]
marks = [40, 50, 60, 70, 85]
plt.scatter(hours, marks)
plt.title('Study Hours vs Marks')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.show()
```

```python
# labels and legend - explain what each line represents
import matplotlib.pyplot as plt
months = ['Jan', 'Feb', 'Mar']
shop_a = [100, 120, 140]
shop_b = [90, 110, 130]
plt.plot(months, shop_a, label='Shop A')
plt.plot(months, shop_b, label='Shop B')
plt.title('Sales Comparison')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.legend()
plt.show()
```

```python
# simple trend - compare starting and ending sales
sales = [100, 120, 140, 160]
if sales[-1] > sales[0]:
    print('Overall sales trend is increasing')
else:
    print('Overall sales trend is not increasing')
```

## Concepts and Output Explanations

**1. What is Data Visualization?**  
Data visualization means showing data using graphs or charts so that it is easier to understand.

**2. What is Matplotlib?**  
Matplotlib is a Python library used to create charts. We use `matplotlib.pyplot` to draw graphs.

**3. What is a Line Plot?**  
A line plot shows how values change, often over time. **Example:** Sales from January to April. **Output:** The line rises overall from 100 to 180, with a small drop in March.

**4. What is a Bar Chart?**  
A bar chart compares values in different groups. **Example:** Sales of books, pens and bags. **Output:** Pens have the highest sales (80).

**5. What is a Histogram?**  
A histogram groups numerical values into ranges and shows their frequencies. **Example:** Student marks. **Output:** Taller bars mean more marks fall in that range.

**6. What is a Scatter Plot?**  
A scatter plot shows the relationship between two numerical values. **Example:** Study hours and marks. **Output:** In this sample, more study hours are associated with higher marks.

**7. What are Title, Labels and Legend?**  
A **title** names the chart. **X-axis and Y-axis labels** explain the values shown. A **legend** identifies different lines or groups. **Output:** The two lines are named Shop A and Shop B.

**8. What is a Trend?**  
A trend is the general direction of values. **Example:** Sales of 100, 120, 140, 160. **Output:** `Overall sales trend is increasing`.

**Summary:** Line plots show change, bar charts compare groups, histograms show distributions, and scatter plots show relationships.
