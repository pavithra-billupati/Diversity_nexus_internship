```python
def Number(*num):
    #pos,nag and zeros
    positive=""
    negative=""
    count=0
    for i in num:
        if i>0:
            positive +=str(i)+" "
        elif i<0:
            negative +=str(i)+" "
        else:
            count+=1
    print(f" positive numbers {positive}")
    print(f" negative numbers {negative}")
    print(f"zero are {count}")
Number(9,8,78,65,89,37,0,-3,98)
```

```python
def even_odd(*numbers):
    even=""
    odd=""
    for i in num:
        if i%2 ==0:
            even +=str(i)+" "
        else:
            odd +=str(i)+" "
    print(f"even numbers are {even}")
    print(f" odd numbers are {odd}")
even_odd(67,90,76,9,0,4,8,-9)
```

```python
def prime(number):
    for i in (2,number):
        if number%i==0:
            return "not a prime"
    return "it is prime"
prime_result=prime(79)
print(prime_result)
```

```python
def calculate_grade(marks):
    if marks<=100 and marks>=0:
        if marks>=90 and marks<=100:
            return "A 100-90"
        elif marks>=75 and marks<=89:
            return "B 89-70"
        elif marks>=60 and marks<=74:
            return "C 74-60"
        elif marks>=40 and marks<=59:
            return "D 59-40"
        else:
            return "F below 40"
    else:
        return "Invalid marks"
    
grade_result=calculate_grade(-9)
print(grade_result)
```

```python
def is_palindrome(value):
    if type(value)==int:
        num=value
        r_value=0
        while num!=0:
            digit=num%10
            r_value=(r_value*10)+digit
            num=num//10
        if value==r_value:
            return "is a palindrome"
        else:
            return "not a palindrome"
    elif type(value)==str:
        re_value=value[::-1]
        if value==re_value:
            return "is palindrome"
        else:
            return "not a palindrome"
palindrome_result=is_palindrome("paap")
print(palindrome_result)
```

```python
def calculate(operation, *numbers):
    if operation=="add":
        return sum(*numbers)
    elif operation=="multiply":
        mul=1
        for i in numbers:
            mul=mul*i
        return mul
    elif operation=="maximum":
        return max(numbers)
    elif operation=="minimum":
        return min(numbers)
    else:
        return "invalid"

calculate_result=calculate("minimum",9,8,89,76,45,34,12)
print(calculate_result)
```
