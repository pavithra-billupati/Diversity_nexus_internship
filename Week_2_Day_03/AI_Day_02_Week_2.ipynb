```python
# Main Hands-on Activity: Data Processing Script
# Build a simple data processing script that reads student records, cleans the data, handles bad records, and exports the cleaned result.
# Use a CSV file such as:
# name,age,marks,city
# Rahul,21,85,Hyderabad
# Priya,22,92,Bangalore
# ,20,75,Chennai
# Arjun,abc,80,Chennai
# Sneha,21,95,
# Kiran,20,88,Hyderabad
import csv

with open("students.csv", "w", newline="") as file:
    data = csv.writer(file)

    data.writerows([
        ["name", "age", "marks", "city"],
        ["Rahul", 21, 85, "Hyderabad"],
        ["Priya", 22, 92, "Bangalore"],
        ["", 20, 75, "Chennai"],
        ["Arjun", "abc", 80, "Chennai"],
        ["Sneha", 21, 95, ""],
        ["Kiran", 20, 88, "Hyderabad"]
    ])
# Your program should:
# Read the CSV file.
with open("students.csv", "r") as file:
    data = csv.DictReader(file)
    for row in data:
        print(row)
# Check for missing names or cities.
with open("students.csv", "r") as file:
    data = csv.DictReader(file)
    valid_students = []
    invalid_students = []
    for row in data:
        if row["name"] == "" or row["city"] == "":
            print("Invalid record:", row)
            invalid_students.append(row)
            continue
# Handle invalid age/marks values without crashing.
        try:
            age=int(row["age"])
            marks=int(row["marks"])
        except ValueError:
            print("Invalid age or marks:",row)
            invalid_students.append(row)
            continue
# Clean valid records.
        student = {
            "name": row["name"].strip(),
            "age": age,
            "marks": marks,
            "city": row["city"].strip()
        }

        valid_students.append(student)
# Ignore or separately report invalid records.
# Calculate the average marks of valid students.
        total = 0
        l1=[]
        for student in valid_students:
            total += student["marks"]
            l1.append(student["marks"])
        average = total / len(valid_students)
        print("Average marks:", average)
# Find the highest scorer.
        print("The highest scorer:",max(l1))
# Export valid records into cleaned_students.csv.
with open("cleaned_students.csv", "w", newline="") as file:
    fieldnames = ["name", "age", "marks", "city"]
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(valid_students)
# Save invalid records into errors.csv or display them clearly.
with open("errors.csv", "w", newline="") as file:
    fieldnames = ["name", "age", "marks", "city"]
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(invalid_students)
```
